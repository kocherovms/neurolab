```python
project_root_dir = !git rev-parse --show-toplevel
project_root_dir = project_root_dir[0]

import os
import sys
import io
import pickle

sys.path.append(os.path.join(project_root_dir, 'lib'))
import launch_dispatcher
```

```python
script_code = '''
import os
import torch
f = open('/script/output.txt', 'wt')
f.write(f'Hello, world! Current working directory is "{os.getcwd()}"')
print(f'{torch.cuda.is_available()=}')

'''

launch_request = dict(
    script_name='test_script',
    script_version=1,
    script_code=script_code,
    result_fname='/script/output.txt',
    keep_container=False,
    return_logs=True,
)
launch_result_metadata, launch_result_body = launch_dispatcher.LaunchRequest.run(launch_request)
```

```python
with io.BytesIO(launch_result_body) as b:
    unpickled = pickle.load(b)
    result_file = unpickled['result_file']
    logs = unpickled['logs']
```
