# Goal

Выяснилось, что агенты отвратительно играют на третьем уровне. Теряют жизни за просто так, тогда как их можно было бы сэкономить и продвинуться дальше. И это наблюдается в любых линейках.

Тут проверяем: а можно ли научить агента проходить 3-ий уровень без потерь жизни. Ведь 1-ый и 2-ый уровни он проходит без потерь. Игнорируем факт, что агент может разучиться играть на других уровнях - хочется просто посмотреть да-нет.

Исследования, которые были посвящены этой проблеме:
- `18n_study_40.3f`
- `18n_study_40.3g`
- `18n_study_48.3c`
- `18n_study_48.3d`

Видно, что 40-ая линейка показывает более быстрое и обильное схватывание, чем 48-ая: `40.3f` дал пять агентов, а `40.3g` - четыре агента, которые умеют проходить 3-ий уровень на одной жизни, тогда как `48.3c` и `48.3d` - всего по одному жидкому прострелу. При этом особой разницы между `40.3f` и `40.3g` не видно, что заставляет думать о том, что родители более удачные, чем у 48-ой линейки.

Кажется, что `48.3d` чуть получше, чем `48.3c`, так как `sequence_ages` короче и `batch_size` больше. 

Ожидаемо, что все исследования показывают деградацию по `levels_passed` - агенты на голой диете 3-го уровня разучиваются играть в первые. 

Заметил интересное явление. Если посмотреть видосы, как играют чемпионы этих исследований, то кажется, что 48-ая линейка более подвержена небрежным шагам агента. Это когда, например, агент стоит на нижнем ряду льдин, и над ним проплывают две льдины с небольшим фазовым смещением относительно нижней: хочешь вверх и вправо, хочешь вверх и влево прыгай - не промахнёшся. Но агент прыгает аккурат вверх и плюхается в воду. Или просто агент стоя на льдине ни с того, ни с сего делает несколько шагов и соскальзывает в воду. Возможно дрифт батч-норма заставляет агента в 40-ой линейке выучивать более надёжные позиции и шаги, т.к. для него, возможно, края льдин расплываются. А в 48-ой все чётче, но и тем самым опаснее что-ли. Короче феномен есть...


**Выводы**
1) главный вывод для любой линейки: агент может научиться проходить 3-ий уровень без потери жизни. Это факт. 
2) можно заняться диетологией и подмешать 3-ий уровень в последней на третий шаг. 
3) а можно начать думать в сторону более продвинутого action planning. Чтобы агент отлавливал сложные моменты и фокусно в этих сложных моментах генерил новые ходы. Сейчас текущий экшн планинг - это первый попашийся и самый дуболомный вариант. Надо двигаться в сторону модулированного экшн планинга, когда есть какой-то замысел, и экшн-план есть производная от этого замысла
4) 10 шагов вперёд в экшн планинге - это чуть больше пол-секунды игрового времени. Агент не сможет думать даже тактически. Типа "постою на льдине, подожду, когда соседний ряд обернётся" - пол-секунды не позволяют такое продумать. Горизонт планирование должен быть больше.
